```python
tooltip = '클릭하세요.'
tooltips = [[37.5678264, 126.9776579], [37.565826456, 126.9796579]]
lat = 37.56672955126769
lon = 126.97839477762068
popups =['aa', 'bb']
latlngs = [[37.5678264, 126.9776579], [37.565826456, 126.9796579]]
m = folium.Map(location=[lat, lon],tiles='OpenStreetMap', zoom_start=15)
folium.Marker([lat, lon], popup='서울시청', tooltip=tooltip).add_to(m)
for i in range(len(latlngs)):
    folium.Marker([latlngs[i][0], latlngs[i][1]], 
                  popup=popups[i], tooltip=f'{tooltips[i][0]}, {tooltips[i][1]}').add_to(m)
m
```

```python
N=100
latitudes = np.random.uniform(low=35, high=39, size=N)
longitudes = np.random.uniform(low=125, high=129, size=N)

# (N, 2) 형태로 결합
data = np.column_stack((latitudes, longitudes))
popups = [str(i) for i in range(N)]
print(data)
```

```python
m = folium.Map(location=[lat, lon], zoom_start=15)
for i in range(N):
    folium.Marker([data[i][0], data[i][1]], 
                  popup=f'위도: {data[i][0]}, 경도: {data[i][1]}',
                  icon=folium.Icon(color='green', icon='info-sign'),
                  tooltip=f'{data[i][0]}, {data[i][1]}').add_to(m)
m
```

```python
from folium import plugins
m = folium.Map(location=[lat, lon], zoom_start=10)
plugins.MarkerCluster(data, popups=popups).add_to(m)
m
```

```python
# 부산진구 CCTV 설치 위치 마크 표시
df = pd.read_csv('data8/CCTV_20190917.csv')
df
```

```python
df1 = df.loc[df['카메라대수']>=1, :]
df1
```

```python
popups = df1['소재지도로명주소'].tolist()
popups
```

```python
# 중심 위도, 경도 계산
lat_avg = np.array(df1['위도'].tolist()).mean()
lng_avg = np.array(df1['경도'].tolist()).mean()
lat_avg, lng_avg
m = folium.Map(location=[lat_avg, lng_avg], zoom_start=15)
```

```python
data = np.array([df1['위도'].tolist(), df1['경도'].tolist()]).T
for i in range(len(data)):
    folium.Marker([data[i][0], data[i][1]], 
                  popup=popups[i], 
                  icon=folium.Icon(color='red', icon='info-sign'),
                  tooltip=f'{data[i][0]}, {data[i][1]}').add_to(m)
m
```

```python
m = folium.Map(location=[lat_avg, lng_avg], zoom_start=12)
plugins.MarkerCluster(data, popups=popups).add_to(m)
m.save('data8/cctv_map.html')
m
```
