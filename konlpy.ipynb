```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from wordcloud import WordCloud
from collections import Counter
```

```python
df = pd.read_table('data/ratings_train.txt')
df.head()
```

```python
df.isnull().sum()
```

```python
df = df.dropna(how='any')
df.isnull().sum()
```

```python
# 한글 빼고는 모두 공백으로 처리시킴
df['document'] = df['document'].str.replace("[^ㄱ-ㅎㅏ-ㅣ가-힣]", "")
df
```

```python
temp_list=[]
for sentence in df['document']:
    s_list = okt.pos(sentence)
    for word, tag in s_list:
        if tag in ['Noun', 'Adjective']:
            temp_list.append(word)

temp_list
```

```python
len(temp_list)
```

```python
count = Counter(temp_list)
count2 = count.most_common(100)
count2
```

```python
from konlpy.tag import *
okt = Okt()
```

```python
im = Image.open('data/img01.jpg')
mask_arr = np.array(im)
```

```python
from wordcloud import WordCloud
wc = WordCloud(width=1200, height=1000, background_color='white', 
               font_path=font_path, random_state=0, colormap='Accent_r', mask=mask_arr)
plt.imshow(wc.generate_from_frequencies(dict(count2)))
plt.axis('off')
plt.savefig('data/wc_movie.png')
plt.show()
```

```python
list=[]
stopwords = ['의', '가', '이', '은', '들', '는', '을', '를', '으로', '자',  '와', '한', '하다', 
             '그', '저', '것', '수', '있', '없' '다', '되', '게', '고', '으로', '로', '에서', '부터', '까지',
             '때문', '때', '안', '잘', '많이', '좀', '더', '더욱', '또한', '하지만', '그래서', '그러나', '그런데']
for sentence in df['document']:
    s_list = okt.pos(sentence)
    for word, tag in s_list:
        if tag in ['Noun', 'Adjective'] and word not in stopwords:
            list.append(word)
count = Counter(list)
count2 = count.most_common(100)
wc = WordCloud(width=1200, height=1000, background_color='white', 
               font_path=font_path, random_state=0, colormap='Accent_r', mask=mask_arr)
plt.imshow(wc.generate_from_frequencies(dict(count2)))
plt.axis('off')
plt.savefig('data/wc_movie2.png')
plt.show()
```
