```python
emma_raw = nltk.corpus.gutenberg.raw('austen-emma.txt')
emma_raw
```

```python
from nltk.tokenize import word_tokenize
text = "Hello, how are you?"
tokens = word_tokenize(text)
print(tokens)
```

```python
from nltk.tokenize import sent_tokenize
print(sent_tokenize(emma_raw)[10])
```

```python
from nltk.tokenize import word_tokenize

print(word_tokenize(emma_raw))
```

```python
from nltk.tokenize import RegexpTokenizer

#word가 1자 이상인것만 골라서 나누겠다.(대괄호 안에 \w가 워드가 1글자란 의미인거 같고, +가 이상을 의미하는것 같음.)
regTokenize = RegexpTokenizer("[\w]+")

print(regTokenize.tokenize(emma_raw))
```

```python
from nltk.stem import PorterStemmer, LancasterStemmer
st1 = PorterStemmer()
st2 = LancasterStemmer()
words=['fly', 'flies', 'flying', 'flew', 'flown']
print('Poster Stemmer', [st1.stem(w) for w in words])
print('Lancster Stemmer', [st2.stem(w) for w in words])
```

```python
from nltk.stem import WordNetLemmatizer
lm=WordNetLemmatizer()
v1 = [lm.lemmatize(w, 'v') for w in words]
print(v1)
```

```python
sentence = emma_raw[:1000]
sentence
```

```python
from nltk.tag import pos_tag
tagged_list = pos_tag(regTokenize.tokenize(emma_raw))
tagged_list
```

```python
nltk.help.upenn_tagset('NNP')
```

```python
# 첫글자가 대문자면 고유명사로 인식해버림
noun_list = [t[0] for t in tagged_list if t[1] == 'NNP']
noun_list
```

```python
stop_words = ['Mr.', 'Mrs', 'Miss', 'Mr', 'Mrs', 'Dear', 'A', 'No', 'Ah', 'Oh', 'VOLUME', 'Between', 'Woodhouse']
name_list = [t[0] for t in tagged_list if t[1]=='NNP' and t[0] not in stop_words]
name_list
```

```python
fd_name = nltk.FreqDist(name_list)
fd_name
```

'whitman_leaves.txt'를 다음을 수행하라
1. 워드(regexp totkenize) 토큰을 구하라.
2. 품사 태깅을 수행한다.
3. 품사 태깅 리스트에서 고유 명사를 추출하라.
4. 고유 명사리스트에서 출현빈도를 구하라.
5. 불용어 제거 후 출현빈도를 구하라.

```python
whitman_leaves = nltk.corpus.gutenberg.raw('whitman-leaves.txt')
whitman_leaves
```

```python
from nltk.tokenize import word_tokenize

print(word_tokenize(whitman_leaves))
```

```python
from nltk.tokenize import RegexpTokenizer

regTokenize = RegexpTokenizer("[\w]+")
print(regTokenize.tokenize(whitman_leaves))
```

```python
from nltk.tag import pos_tag
tagged_list1 = pos_tag(regTokenize.tokenize(whitman_leaves))
tagged_list1
```

```python
noun_list1 = [t[0] for t in tagged_list1 if t[1] == 'NNP']
noun_list1
```

```python
fd_name1 = nltk.FreqDist(noun_list1)
fd_name1
```

```python
stop_words1 = ['O', 'Come', 'En', 'Body', 'Man', 'Life', 'Returning', 'Soul', 'Self', 'A', 'My']
whitman_leaves_nnp_list = [t[0] for t in tagged_list1 if t[1]=='NNP' and t[0] not in stop_words1]
whitman_leaves_nnp_list
```

```python
fd_name1 = nltk.FreqDist(whitman_leaves_nnp_list)
fd_name1
```
