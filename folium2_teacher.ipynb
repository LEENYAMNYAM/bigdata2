```python
happy_만족도=pd.read_excel('data4/삶의만족도.xlsx')
happy_만족도.shape
happy_건강=pd.read_excel('data4/건강.xlsx')
happy_건강.shape
happy_경제=pd.read_excel('data4/경제.xlsx')
happy_경제.shape
happy_사회참여=pd.read_excel('data4/관계및사회참여.xlsx')
happy_사회참여.shape
happy_교육=pd.read_excel('data4/교육.xlsx')
happy_교육.shape
happy_안전=pd.read_excel('data4/안전.xlsx')
happy_안전.shape
happy_여가=pd.read_excel('data4/여가.xlsx')
happy_여가.shape
happy_환경=pd.read_excel('data4/환경.xlsx')
happy_환경.head()
```

```python
happy_df=happy_만족도
happy_df['건강']=happy_건강['평균']
happy_df['경제']=happy_경제['평균']
happy_df['사회참여']=happy_사회참여['평균']
happy_df['교육']=happy_교육['평균']
happy_df['안전']=happy_안전['평균']
happy_df['여가']=happy_여가['평균']
happy_df['환경']=happy_환경['평균']
happy_df.head()
```

```python
happy_group=happy_df.groupby('시도')['삶의 만족도'].mean()
happy_group
```

```python
map=folium.Map(location=[37.559819, 126.96389], zoom_start=7)
folium.Choropleth(geo_data=sido_geo, 
                  data=happy_group,
                  columns=['sido','total'],
                  fill_color='RdYlGn',
                  fill_opacity=0.7,
                  line_opacity=0.1,
                  key_on='properties.CTP_KOR_NM',
                  legend_name='시도별 총 인구수'
                  ).add_to(map)
map
```

```python
map=folium.Map(location=[37.559819, 126.96389], zoom_start=7)
folium.Choropleth(geo_data=sido_geo, 
                  data=happy_group,
                  columns=['sido','삶의 만족도'],
                  fill_color='RdYlGn',
                  fill_opacity=0.7,
                  line_opacity=0.1,
                  key_on='properties.CTP_KOR_NM',
                  legend_name='시도별 총 인구수'
                  ).add_to(map)
folium.GeoJson(
    sido_geo,
    name='지역 이름',
    tooltip=folium.GeoJsonTooltip(fields=['CTP_KOR_NM'], aliases=['구군:'],localize=True,
        sticky=False)
).add_to(map)
map
```

```python
happy_group1=happy_df.groupby(['시도', '구군'])['삶의 만족도'].mean()
happy_group1
```

```python
happy_group2=happy_group1.reset_index()
happy_group2
```

```python
map=folium.Map(location=[37.559819, 126.96389], zoom_start=7)
folium.Choropleth(geo_data=sig_geo, 
                  data=happy_group2,
                  columns=['구군','삶의 만족도'],
                  fill_color='RdYlGn',
                  fill_opacity=0.7,
                  line_opacity=0.1,
                  key_on='properties.SIG_KOR_NM',
                  legend_name='시도별 총 인구수'
                  ).add_to(map)

for feature in sig_geo['features']:
    sig_name = feature['properties']['SIG_KOR_NM']
    value_row = happy_group2[happy_group2['구군'] == sig_name]
    if not value_row.empty:
        feature['properties']['삶의 만족도'] = value_row['삶의 만족도'].values[0]
    else:
        feature['properties']['삶의 만족도'] = 'N/A'
folium.GeoJson(
    sig_geo,
    name='지역 이름',
    tooltip=folium.GeoJsonTooltip(fields=['SIG_KOR_NM','삶의 만족도'], aliases=['구군:', '삶의 만족도:'],localize=True,
        sticky=False)
).add_to(map)
map
```
